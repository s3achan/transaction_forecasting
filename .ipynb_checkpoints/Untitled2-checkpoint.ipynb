```python
# Transaction + FX Forecasting Project
# Author: Shikshya Bhattachan
# GitHub: github.com/s3achan

# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from prophet import Prophet

# 2. Load synthetic transaction dataset
transactions = pd.read_csv("synthetic_transaction_data.csv", parse_dates=["date"])
print(transactions.head())

# 3. Exploratory Data Analysis
corridor = "USD/INR"
df_corridor = transactions[transactions["corridor"] == corridor]

plt.figure(figsize=(10,5))
plt.plot(df_corridor["date"], df_corridor["net_revenue"])
plt.title(f"Historical Net Revenue - {corridor}")
plt.xlabel("Date")
plt.ylabel("Net Revenue (USD)")
plt.show()

# 4. Prepare data for forecasting
df = df_corridor.rename(columns={"date":"ds","net_revenue":"y"})

# 5. Prophet Forecasting Model
model = Prophet(yearly_seasonality=True, daily_seasonality=False)
model.fit(df)

future = model.make_future_dataframe(periods=365*5)  # forecast 5 years ahead
forecast = model.predict(future)

# 6. Plot forecast
fig = model.plot(forecast)
plt.title(f"5-Year Net Revenue Forecast - {corridor}")
plt.show()

# 7. Fee Scenario Analysis
fees = [0.015, 0.02, 0.025]  # 1.5%, 2%, 2.5%
results = {}

for fee in fees:
    forecast[f"net_revenue_{fee}"] = forecast["yhat"] * (1 - fee)
    results[fee] = forecast[f"net_revenue_{fee}"].sum()

print("Fee Scenario Results (Total Net Revenue over 5 years):")
for fee, total in results.items():
    print(f"{fee*100:.1f}% fee: ${total:,.0f}")

# 8. Save outputs
forecast.to_csv("forecast_results.csv", index=False)
```
